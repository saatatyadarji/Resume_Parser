```python
import pdfplumber, docx2txt, re

def extract_text_from_pdf(pdf_path):
    text = ""
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            text += page.extract_text() + "\n"
    return text

def clean_text(text):
    text = re.sub(r'\n+', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
import re
def extract_email(text):
  pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
  matches = re.findall(pattern , text)
  return matches
def extract_phone(text):
  pattern = r'(\+?\d{10,15}|\b\d{10}\b)'
  matches = re.findall(pattern , text)
  return matches
def extract_links(text):
  pattern = r'http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\(\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+'
  matches = re.findall(pattern , text)
  return matches
```

```python
def extract_skills(text):
    skill_keywords = [
        'python', 'java', 'c++', 'sql', 'machine learning', 'deep learning','data analysis', 'nlp', 'tensorflow', 'pandas', 'numpy', 'react','C', 'HTML', 'CSS', 'Javascript', 'PHP','MySQL','Linux', 'Cryptography', 'Digital Forensics''excel', 'django', 'flask', 'power bi', 'tableau', 'DAX', 'Cognos Analytics', 'Matplotlib', 'Seaborn' , 'git' , 'hadoop' , 'aws' , 'mongodb','docker','restapi','databricks'

    ]
    found_skills = []
    for skill in skill_keywords:
        if re.search(r'\b' + re.escape(skill) + r'\b', text.lower()):
            found_skills.append(skill)
    return list(set(found_skills))
```

```python
import re
import json
import nltk
import spacy

nltk.download('punkt')
nlp = spacy.load("en_core_web_sm")


def extract_experience(text):
    exp_sections = re.findall(r'(experience|work experience|employment history|internship)(.*?)(education|project|skills|$)',
                              text, re.I | re.S)
    experiences = []
    for section in exp_sections:
        content = section[1]
        sentences = nltk.sent_tokenize(content)
        for s in sentences:
            if re.search(r'\b(intern|developer|engineer|manager|analyst|duration|responsibilit|internship|experience|trainee)\b', s, re.I):
                experiences.append(s.strip())
    return experiences
```

```python
def extract_projects(text):
    proj_sections = re.findall(r'(projects?|internship)(.*?)(education|experience|skills|$)',
                               text, re.I | re.S)
    projects = []
    for section in proj_sections:
        content = section[1]
        sentences = nltk.sent_tokenize(content)
        for s in sentences:
            if re.search(r'(power bi|python|dashboard|analysis|model|eda|classifier|dataset)', s, re.I):
                projects.append(s.strip())
    return projects
```

```python
def extract_education(text):
    edu_sections = re.findall(r'(education|qualification|academic)(.*?)(project|experience|skills|$)',
                              text, re.I | re.S)
    education = []
    for section in edu_sections:
        content = section[1]
        lines = content.split('\n')
        for line in lines:
            if re.search(r'\b(btech|b\.tech|mtech|b\.sc|m\.sc|bachelor|master|university|college|school)\b', line, re.I):
                education.append(line.strip())
    return education
```

```python
def parse_resume(text):
    data = {
        "Email": extract_email(text),
        "Phone": extract_phone(text),
        "Links": extract_links(text),
        "Skills": extract_skills(text),
        "Experience": extract_experience(text),
        "Projects": extract_projects(text),
        "Education": extract_education(text)
    }
    return data
```
